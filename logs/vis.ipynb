```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# 로그 파일 경로
norm_path = "./t1norm_log.csv"
pred_path = "./pred_log.csv"

# ----------- Load Data -----------
if not os.path.exists(norm_path) or not os.path.exists(pred_path):
    raise FileNotFoundError("로그 파일이 존재하지 않습니다.")

norm_df = pd.read_csv(norm_path)
pred_df = pd.read_csv(pred_path)

# ----------- Visualization 1: Norm -----------

norm_path = "./t1norm_log.csv"
norm_df = pd.read_csv(norm_path)
plt.figure(figsize=(10, 6))
for emb_name in norm_df.columns[1:]:
    plt.plot(norm_df["layer"], norm_df[emb_name], label=emb_name, marker='o')
plt.title("Layer-wise Embedding Norm")
plt.xlabel("Layer")
plt.ylabel("L2 Norm")
plt.legend()
plt.grid(True)
plt.show()

norm_path = "./t2norm_log.csv"
norm_df = pd.read_csv(norm_path)
plt.figure(figsize=(10, 6))
for emb_name in norm_df.columns[1:]:
    plt.plot(norm_df["layer"], norm_df[emb_name], label=emb_name, marker='o')
plt.title("Layer-wise Embedding Norm")
plt.xlabel("Layer")
plt.ylabel("L2 Norm")
plt.legend()
plt.grid(True)
plt.show()

norm_path = "./ydnorm_log.csv"
norm_df = pd.read_csv(norm_path)
plt.figure(figsize=(10, 6))
for emb_name in norm_df.columns[1:]:
    plt.plot(norm_df["layer"], norm_df[emb_name], label=emb_name, marker='o')
plt.title("Layer-wise Embedding Norm")
plt.xlabel("Layer")
plt.ylabel("L2 Norm")
plt.legend()
plt.grid(True)
plt.show()

# ----------- Visualization 2: Prediction Mean/Std -----------

fig, ax1 = plt.subplots(figsize=(10, 6))


# Mean plot
color_mean = 'tab:blue'
ax1.set_xlabel('Layer')
ax1.set_ylabel('Mean of Prediction', color=color_mean)
ax1.plot(pred_df['layer'], pred_df['mean'], color=color_mean, marker='o', label='Mean')
ax1.tick_params(axis='y', labelcolor=color_mean)
ax1.set_ylim(0, 1.0)

# Twin axis for std
ax2 = ax1.twinx()
color_std = 'tab:red'
ax2.set_ylabel('Std of Prediction', color=color_std)
ax2.plot(pred_df['layer'], pred_df['std'], color=color_std, marker='x', linestyle='--', label='Std')
ax2.tick_params(axis='y', labelcolor=color_std)
ax2.set_ylim(0, 0.5)

# Title and grid
plt.title("Prediction Mean and Std per Layer")
fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 로그 파일 경로
pred_path = "pred_log.csv"
pred_df = pd.read_csv(pred_path)

# ---------- Plot 1: Mean ----------
plt.figure(figsize=(10, 5))
plt.plot(pred_df["layer"], pred_df["mean"], marker='o', color='blue', alpha=0.7)
plt.title("Prediction Mean per Layer")
plt.xlabel("Layer")
plt.ylabel("Mean")
plt.ylim(0, 1)
plt.grid(True)
plt.show()

# ---------- Plot 2: Std ----------
plt.figure(figsize=(10, 5))
plt.plot(pred_df["layer"], pred_df["std"], marker='x', linestyle='--', color='red', alpha=0.7)
plt.title("Prediction Std per Layer")
plt.xlabel("Layer")
plt.ylabel("Standard Deviation")
plt.ylim(0, 0.5)
plt.grid(True)
plt.show()
```
